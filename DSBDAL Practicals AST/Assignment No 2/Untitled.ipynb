```python
import pandas as pd
import numpy as np
```

```python
data={
    'Student_ID':range(1,11),  
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva', 'Frank', 'Grace', 'Helen', 'Ian', 'Jack'],
    'Math_Score': [20, 85, 92, 88, np.nan, 76, 102, 55, 60, 95], #102 is outlier and a missing value
    'Science_Score': [72, 80, 78, 85, 90, 70, 89, np.nan, 60, 15],  # 105 is an outlier
     'Attendance_Percentage': [85, 90, 75, 88, 70, 95, 100, 65, 82, 78]
}
```

```python
df=pd.DataFrame(data)
df.to_csv('Academic_performance1.csv',index=False)
```

```python
#load the dataset
df=pd.read_csv('Academic_performance1.csv')
print("Missing Values\n",df.isnull().sum())
```

```python
df.columns = df.columns.str.strip()

# Handle missing values (e.g., fill with mean for numeric columns)
df['Math_Score'] = df['Math_Score'].fillna(df['Math_Score'].mean())
df['Science_Score'] = df['Science_Score'].fillna(df['Science_Score'].mean())

```

```python
df.head(n=10)
```

```python
# Confirm replacement
print("\nAfter handling missing values:\n", df.isnull().sum())
```
